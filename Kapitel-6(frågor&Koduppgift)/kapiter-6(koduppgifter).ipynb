```python
import pandas as pd 
df = pd.read_csv("../kapitel-4(frågor&Koduppgift)/hr_employee_data.csv", sep=";")
print(df.head())
print(df.shape)
```

```python
print(df["satisfaction_level"].head())
print(df["last_evaluation"].head())
print(df.dtypes)
```

```python
df["satisfaction_level"] = (
  df["satisfaction_level"]
  .str.replace("%", "", regex=False)
  .astype(float) /100
)

df["last_evaluation"] = (
  df["last_evaluation"]
  .str.replace("%", "", regex=False)
  .astype(float) /100
)

print(df[["satisfaction_level", "last_evaluation"]].head())
```

```python
X = df[[
    "satisfaction_level",
    "last_evaluation",
    "number_project",
    "average_montly_hours",
    "time_spend_company"
]]

print(X.head())
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled[:5])
```

```python
from sklearn.cluster import KMeans
kmeans = KMeans(
  n_clusters=3,
  random_state=42,
  n_init=10
  
)

df["cluster"] = kmeans.fit_predict(X_scaled)
print(df["cluster"].value_counts().sort_index())
```

```python
cluster_result = df.groupby("cluster")[[
  "satisfaction_level",
  "last_evaluation",
  "number_project",
  "average_montly_hours",
  "time_spend_company"
]].mean()

print(cluster_result.round(2))
```

```python
print(df.groupby("cluster")["left"].mean().round(2))
print(pd.crosstab(df["cluster"], df["left"]))
```

Utifrån resultatet så fick jag tre olika grupper. första klustret 0 hade högst nöjda medarbetare. Kluster 1 hade arbetat mindre timmar och fick lägre resultat i utvärdering. Kluster 2 hade mest jobb och arbetade mestadels timmar, och har även jobbat längst i företaget också. Men de var mindre nöjda om man jämför med de i kluster 0.

detta är en bra ide för företag som ska kunna utvärdera och kolla igenom olika anställda i grupper för att se om de behöver stöd, eller att man kollar de som har för hög arbetsbelastning. Och kan även få en bättre perspektiv hur alla sina medarbetare har det på jobbet.

```python
import pandas as pd 

df = pd.read_csv("1980sClassics.csv")
print(df.head())
print(df.shape)
print(df.columns)
```

```python
X = df [[
  "Danceability",
  "Energy",
  "Loudness",
  "Speechiness",
  "Acousticness",
  "Instrumentalness",
  "Liveness",
  "Valence",
  "Tempo"
]]
print(X.head())
print(X.shape)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled[:5])
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans (
  n_clusters=3,
  random_state=42,
  n_init=10

)

df["cluster"] = kmeans.fit_predict(X_scaled)
print(df["cluster"].value_counts().sort_index())
```

```python
cluster_result = df.groupby("cluster")[[
  "Danceability",
    "Energy",
    "Loudness",
    "Speechiness",
    "Acousticness",
    "Instrumentalness",
    "Liveness",
    "Valence",
    "Tempo"

]].mean()
print(cluster_result.round(2))
```

Jag har använt mig utav K-means för att dela in Spotify låtarna från dataset i tre olika grupper.
jag har valt olika variabler såsom: acoustic, tempo, energy, liveness och m.m, för att se hur lika låtarna är med varandra.

i första klustret (kluster 0) så hade låtarna innehållet med hög loudness och hade ett högt värde med valence.
Vilket innebär att den gruppen är mest glada och är mer utåtrikade till dans.

kluster 1 hade innehåll med lägre energi och högre värde på acoustic. Kluster 2 hade låtar med snabbare tempo 
och högre energi.

Jag tycker att denna klustring är särskilt bra om man vill gruppera specifika musikgenre baserad på ljudet de gör/skapar, och samla de i en spellista för att lägga till olika låtar baserad på stämning och aktiviteter.
